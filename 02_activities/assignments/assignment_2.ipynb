  # Question Two: Path to Leaves

  Given the `root` of a binary tree, return all root to leaf paths in any order.

  ## Examples

  ### Example 1

  ![](./images/q1_ex1.png)

  Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

  ### Example 2

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 7, 8]`

  Output: [[10, 7], [10, 9, 8]]

</details>

#### Starter Code for Question 2

```python
from typing import List, Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def bt_path(root: Optional[TreeNode]) -> List[List[int]]:
    """
    Return all root-to-leaf paths in a binary tree.
    Uses depth-first search (DFS) with backtracking.
    Time:  O(n)   — every node is visited exactly once
    Space: O(h)   — call stack depth equals tree height h;
                    O(log n) for balanced trees, O(n) worst-case (skewed)
    """
    if not root:
        return []

    results = []

    def dfs(node: TreeNode, path: List[int]) -> None:
        path.append(node.val)
        # Leaf node: record the completed path
        if not node.left and not node.right:
            results.append(list(path))   # snapshot the current path
        else:
            if node.left:
                dfs(node.left, path)
            if node.right:
                dfs(node.right, path)
        path.pop()  # backtrack

    dfs(root, [])
    return results


# --- Example 1 ---
#        1
#      /   \
#     2     2
#   / \ / \
#  3  5 6  7
root1 = TreeNode(1,
    TreeNode(2, TreeNode(3), TreeNode(5)),
    TreeNode(2, TreeNode(6), TreeNode(7))
)
print("Example 1:", bt_path(root1))   # [[1,2,3],[1,2,5],[1,2,6],[1,2,7]]

# --- Example 2 ---
#      10
#     /  \
#    9    7
#     \
#      8
root2 = TreeNode(10,
    TreeNode(9, None, TreeNode(8)),
    TreeNode(7)
)
print("Example 2:", bt_path(root2))   # [[10,9,8],[10,7]]

# --- New Example: single-node tree (root is also the only leaf) ---
root3 = TreeNode(42)
print("Single node:", bt_path(root3)) # [[42]]

# --- New Example: right-skewed tree ---
#  1
#   \
#    2
#     \
#      3
root4 = TreeNode(1, None, TreeNode(2, None, TreeNode(3)))
print("Right-skewed:", bt_path(root4)) # [[1,2,3]]
```

# Given a string made up entirely of the characters (, ), {, }, [, and ],
# determine whether the brackets are validly matched.
# "Validly matched" means two things:
#   (a) every opening bracket has exactly one corresponding closing bracket of the same type, and
#   (b) brackets are properly nested — a bracket opened inside another must be closed before the outer one is closed.
# Return True if the string satisfies both conditions, False otherwise.


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


# --- New example I created ---
# Input:  s = ")("
# Output: False
# Reason: The closing bracket comes before any opening bracket, so it can never match.
# This is a useful edge case because the string has equal counts of ( and ), yet is still invalid.

# --- Walkthrough of Lisa's example: s = "{[()()]}" → True ---
#
# We process each character in order, maintaining a stack (open_brackets):
#
#  char  | action                            | stack after
#  ------+-----------------------------------+------------
#  '{'   | opening → push                    | ['{']
#  '['   | opening → push                    | ['{', '[']
#  '('   | opening → push                    | ['{', '[', '(']
#  ')'   | closing → top is '(', '('+')' = '()' ✓ → pop   | ['{', '[']
#  '('   | opening → push                    | ['{', '[', '(']
#  ')'   | closing → top is '(', '()' ✓ → pop             | ['{', '[']
#  ']'   | closing → top is '[', '[]' ✓ → pop             | ['{']
#  '}'   | closing → top is '{', '{}' ✓ → pop             | []
#
# Loop ends. Stack is empty → return True. ✓


-   Copy the solution your partner wrote. 


```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    open_brackets = []
    for char in s:
        if char in ['(', '[', '{']:
            open_brackets.append(char)
        else:
            if (open_brackets[-1] + char) in ['[]','()', '{}']:
                open_brackets.pop()
            else:
                return False
    # Check if open_brackets is empty
    if not open_brackets:
        return True
    else:
        return False
    pass

print(is_valid_brackets('([]{})'))          # True
print(is_valid_brackets('([)]'))            # False
print(is_valid_brackets('()[]{}'))          # True
print(is_valid_brackets('[{]}'))            # False
print(is_valid_brackets('{[()()]}'))        # True
print(is_valid_brackets('{()[]'))           # False
print(is_valid_brackets('('))               # False
print(is_valid_brackets('(()'))             # False
```


-   Explain why their solution works in your own words.


# A stack (list used as LIFO) is the ideal data structure here because bracket nesting is
# inherently last-in-first-out: the most recently opened bracket must be the next one closed.
#
# The algorithm:
#   1. Iterate through each character in the string.
#   2. If it is an opening bracket, push it onto the stack.
#   3. If it is a closing bracket, peek at the top of the stack (open_brackets[-1]).
#      Concatenate the top element with the current character and check whether that
#      two-character string is one of the valid pairs '()', '[]', '{}'.
#        - If it matches, the innermost open bracket is now closed → pop it off the stack.
#        - If it does not match, the nesting order is wrong → return False immediately.
#   4. After the loop, check whether the stack is empty.
#      An empty stack means every opening bracket was eventually matched and closed → True.
#      A non-empty stack means at least one opening bracket was never closed → False.


-   Explain the problem’s time and space complexity in your own words.


# Time complexity: O(n)
#   The loop visits each character in the string exactly once.
#   All operations inside the loop — append, pop, index lookup, string concatenation on
#   fixed-length (1-character) strings, and membership check in a fixed-length list — are O(1).
#   So the overall time is linear in the length of the input string.
#
# Space complexity: O(n)
#   In the worst case (e.g. the string is all opening brackets like "((((("),
#   every character is pushed onto the stack, giving O(n) auxiliary space.
#   Note: Lisa wrote this as O(n+1), but the constant is conventionally dropped,
#   so it is simply O(n).

My initial instinct was to reach for a breadth-first search, since that mirrors the level-order representation used in the examples. However, I quickly realized that DFS with backtracking is a better fit: BFS would require storing an entire path alongside every node in the queue, whereas DFS naturally tracks the current path on the call stack and backtracks by popping the path list as recursion unwinds. Choosing the right traversal strategy was the key insight, and working through concrete examples on paper before coding made the implementation straightforward.
